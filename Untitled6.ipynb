```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_squared_error

# Фіксуємо seed для відтворюваності результатів
np.random.seed(42)

# Генеруємо 1000 випадкових значень X (наприклад, від 0 до 10)
X = np.sort(10 * np.random.rand(1000, 1), axis=0)

# Генеруємо цільову змінну Y як функцію від X + випадковий шум
y = np.sin(X).ravel() + np.random.normal(0, 0.2, X.shape[0])

print(f"Згенеровано записів: {X.shape[0]}")
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(f"Середнє після нормалізації: {X_scaled.mean():.4f}")
print(f"Дисперсія після нормалізації: {X_scaled.std():.4f}")
```

```python
# Розбиття на Train (80%) та Test (20%)
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

print(f"Розмір навчальної вибірки: {X_train.shape[0]}")
print(f"Розмір тестової вибірки: {X_test.shape[0]}")
```

```python
k_values = range(1, 51)
mse_scores = []

best_k = 1
best_mse = float('inf')
best_knn_model = None

# Перебираємо різні значення K
for k in k_values:
    # Створюємо та навчаємо модель
    knn = KNeighborsRegressor(n_neighbors=k)
    knn.fit(X_train, y_train)

    # Робимо прогноз на тестових даних
    y_pred = knn.predict(X_test)

    # Оцінюємо якість (Похибка MSE)
    mse = mean_squared_error(y_test, y_pred)
    mse_scores.append(mse)

    # Зберігаємо найкращу модель
    if mse < best_mse:
        best_mse = mse
        best_k = k
        best_knn_model = knn

print(f"Найкраще значення K: {best_k}")
print(f"Мінімальна похибка MSE на тестовій вибірці: {best_mse:.4f}")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# 1. Графік залежності MSE від K
axes[0].plot(k_values, mse_scores, marker='o', color='purple', linestyle='dashed')
axes[0].axvline(x=best_k, color='red', label=f'Найкраще K={best_k}')
axes[0].set_title("Залежність похибки MSE від кількості сусідів (K)")
axes[0].set_xlabel("Кількість сусідів (K)")
axes[0].set_ylabel("Похибка (MSE)")
axes[0].legend()
axes[0].grid(True)

# 2. Графік результатів найкращої регресії
# Генеруємо щільну сітку точок для плавної кривої
X_grid = np.linspace(X.min(), X.max(), 500).reshape(-1, 1)
X_grid_scaled = scaler.transform(X_grid)
y_grid_pred = best_knn_model.predict(X_grid_scaled)

axes[1].scatter(X, y, color='gray', alpha=0.3, s=15, label='Фактичні дані')
axes[1].plot(X_grid, y_grid_pred, color='red', linewidth=3, label=f'KNN Регресія (K={best_k})')
axes[1].set_title(f"Результат KNN-регресії при оптимальному K={best_k}")
axes[1].set_xlabel("X (Ознака)")
axes[1].set_ylabel("Y (Цільова змінна)")
axes[1].legend()

plt.tight_layout()
plt.show()
```
